# Functions

```python
# introduces the function that adds one to a number
#Creating the function
def add_one(number):
  return number + 1
```

```python
#Using the function or calling the function
add_one(3)  # calls the add_one function with 3 as the input
```

```python
# defines function that classifies the bp
def bp_check(systolic_bp):  # defines function to slassify the bp
  if systolic_bp > 120:  # check if bp is above 120
    bp_result = "High"
  else:
    bp_result = "Normal"
  return bp_result  # return bp to classification
```

```python
# tests bp classification function
bp_check(150)
```

# Loops

```python
# creates a list of bp value
bp_list = [120, 135, 170, 110, 100]
```

```python
# loops through each bp value and classifies it
for bp in bp_list:
  print(bp_check(bp))
```

```python
# defining function that classifies bmi
def bmi_check(bmi):
  if bmi < 18.5:  # classify based on its numerical range
    bmi_result = "Underweight"
  elif bmi < 25:
    bmi_result = "Normal"
  elif bmi < 30:
    bmi_result = "Overweight"
  else:
    bmi_result = "Obese"
  return bmi_result  # return bmi classification
```

```python
# tests bmi classification with 24.9
bmi_check(24.9)
```

```python
# defines a function to calculate the bmi
#Group 2
def calculate_bmi(height, weight):
  bmi = weight/height ** 2  # calculated as weight divided by height squared
  return bmi
```

```python
# calculated bmi using the height and weight as positional
calculate_bmi(1.2,57)  # calculates the bmi using named arguments for height and weight
```

```python
calculate_bmi(height = 1.2, weight = 57)
```
