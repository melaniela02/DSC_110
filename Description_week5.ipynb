Treating missingness
1- dropping the rows with missing values
2- fill in with the mean


```python
#drop all rows that have any missing values
df_complete = df.dropna()
```

```python
df_complete.shape[0]
```

```python
df_complete.isna().mean()*100
```

```python
#dropping only rows where select fields have missing value
df_complete_bmi = df.dropna(subset=['bmi'])
```

```python
#verify it
df_complete_bmi['bmi'].isna().sum()
```

```python
#filling na with mean, zero or any other value
#get the mean and assign it to a variable
mean_bp = df['systolic_bp'].mean()
```

```python
print(round(mean_bp,2)) #rounded to two decimals
```

```python
df['systolic_bp'] = df['systolic_bp'].fillna(mean_bp)
```

```python
#verify that the inputation/fillna worked
df['systolic_bp'].isna().sum()
```

Transformation variables

```python
def bmi_check(bmi):
  if bmi < 18.5:
    bmi_result = "underweight"
  elif bmi < 25:
    bmi_result = "normal"
  elif bmi < 30:
    bmi_result = "overweight"
  else:
    bmi_result = "obese"
  return bmi_result
```

```python
df['bmi_category'] = df['bmi'].apply(bmi_check)
```

```python
#verify it
#value counts
df['bmi_category'].value_counts()
```

```python
#new function called cut
df['bmi_category2'] = pd.cut(df['bmi'],
                             bins = [0, 18.5, 25, 30, 100],
                             labels = ["underweight","normal","overweight","obese"])
```

```python
#verifying it
df['bmi_category2'].value_counts()
```

```python
df['bmi_category2'].value_counts(dropna = False) #including even false
```
